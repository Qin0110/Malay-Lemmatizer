```python
import pandas as pd

file = "malindo_dic_2024.tsv"
column = [
    
    "id",
    "root",          # Akar
    "surface",       # Bentuk lahir
    "prefix",        # Awalan/proklitik
    "suffix",        # Akhiran/enklitik
    "circumfix",     # Apitan
    "reduplication", # Penggandaan
    "source",        # Sumber
    "stem",          # Dasar
    "lemma",         # Lema
]
df = pd.read_csv(
    file, 
    sep = "\t",
    names = column,
    encoding = "utf-8"
)

df.head(5)
```

```python
words = input(str("Words: "))

## Words Error checking
output = df.loc[df["surface"] == words.lower()]
output
```

```python
import re

sentences = input(str("Sentences: "))
list_of_words = []
output = [] # need to return array of array

def tokenize(sentences):
    result = re.sub(r'[^\w\s]', '', sentences)
    result = re.split(r'\s+', result)
    return result # return array

for word in tokenize(sentences):
    grammar = df.loc[df["surface"] == word.lower()] # return dataframe
    list_of_words.append(grammar)

for item in list_of_words:
    output.append({
        "Word": item["surface"].to_string(index=False),
        "Root": item["root"].to_string(index=False),
        "Prefix": item["prefix"].to_string(index=False),
        "Suffix": item["suffix"].to_string(index=False),
        "circumfix": item["circumfix"].to_string(index=False),
        "Lemma": item["lemma"].to_string(index=False),
    })
    
print(type(output))
for i in range(len(output)):
    print(output[i])
```
